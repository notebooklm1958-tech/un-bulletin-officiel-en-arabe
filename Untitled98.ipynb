```python
import os, re, requests
from datetime import datetime

H = {"User-Agent": "Mozilla/5.0"}
num = input("Numéro du bulletin officiel (ex: 7533) : ").strip()
dossier = input("Dossier d'enregistrement (Entrée = dossier actuel) : ").strip() or "."

if not num.isdigit():
    raise SystemExit("Numéro invalide.")

# Identifiant de dossier détecté depuis la page (3111 par défaut)
try:
    page = requests.get("https://www.sgg.gov.ma/arabe/BulletinOfficiel.aspx", headers=H, timeout=30).text
    fid = (re.search(r"/BO/AR/(\d+)/", page) or [None, "3111"])[1]
except requests.RequestException:
    fid = "3111"

# Années triées de la plus probable (estimation selon le numéro) à la moins probable
now = datetime.now().year
est = max(1912, min(now, round(2023 + (int(num) - 7194) / 130)))

for y in sorted(range(1912, now + 1), key=lambda y: abs(y - est)):
    url = f"https://www.sgg.gov.ma/BO/AR/{fid}/{y}/BO_{num}_Ar.pdf"
    try:
        r = requests.get(url, headers=H, timeout=60)
    except requests.RequestException:
        continue
    if r.status_code == 200 and r.content[:4] == b"%PDF":
        os.makedirs(dossier, exist_ok=True)
        path = os.path.abspath(os.path.join(dossier, f"BO_{num}_Ar.pdf"))
        open(path, "wb").write(r.content)
        raise SystemExit(f"Téléchargé : {url}\nEnregistré : {path}")

print("Bulletin introuvable.")
```
